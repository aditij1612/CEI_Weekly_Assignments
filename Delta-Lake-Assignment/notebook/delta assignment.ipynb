```python
# Loading dataset 1
df = spark.read.csv(
    "/Volumes/workspace/default/assignment_data/dataset_1.csv",
    header=True,
    inferSchema=True
)

display(df)
```

```python
# checking dataset
print("Total Records:", df.count())
print("Total Columns:", len(df.columns))
```

```python
#finding null values
from pyspark.sql.functions import col
null_df = df.filter(
    col("first_name").isNull() |
    col("email").isNull() |
    col("city").isNull() |
    col("membership").isNull()
)
display(null_df)
```

```python
#finding duplicate rows
duplicate_df = df.groupBy(df.columns).count().filter(col("count") > 1)

display(duplicate_df)
```

```python
# removing null rows
print("Total Records before:", df.count())
clean_df = df.dropna()
print("Total Records after:",clean_df.count())
display(clean_df)
```

```python
#remove duplicates
clean_df = clean_df.dropDuplicates()
print(clean_df.count())
display(clean_df)
```

```python
#saving a delta table
clean_df.write.format("delta") \
.mode("overwrite") \
.saveAsTable("customer_delta")
```

```python
#read a delta table
delta_df = spark.table("customer_delta")
display(delta_df)
```

```python
#uploading second dataset
incremental_df = spark.read.csv(
    "/Volumes/workspace/default/assignment_data/dataset_2.csv",
    header=True,
    inferSchema=True
)
print(incremental_df.count())
display(incremental_df)
```

```python
incremental_df.createOrReplaceTempView("incremental_customer")
```

```python
#merging two datasets so that common rows not repeat and new rows are added
from delta.tables import DeltaTable

deltaTable = DeltaTable.forName(spark, "customer_delta")

(deltaTable.alias("target")
 .merge(
     incremental_df.alias("source"),
     "target.customer_id = source.customer_id"
 )
 .whenMatchedUpdateAll()
 .whenNotMatchedInsertAll()
 .execute())
```

```python
#final merged table
final_df = spark.table("customer_delta")
print("Total Records:", final_df.count())
display(final_df)
```

```python
#checking if there are any duplicates
from pyspark.sql.functions import col
duplicates = final_df.groupBy("customer_id") \
    .count() \
    .filter(col("count") > 1)

display(duplicates)
```
