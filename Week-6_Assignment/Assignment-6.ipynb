Q1: Explain the roles of the Driver, Cluster Manager, and Executor in a Spark application.

ans:Driver runs the main Spark application.It creates the SparkSession.Also converts code into tasks and sends them to executors.

Cluster Manager manages and allocates resources in the cluster.
Starts and monitors executors.

Executors execute tasks assigned by the driver.
Store data in memory and return results to the driver.

Q2: How does Spark’s Lazy Evaluation strategy improve performance when chain-processing large datasets?

ans:Spark does not execute transformations immediately. Instead, it records the transformations and creates a DAG. The actual execution starts only when an action like show() or count() is called.
It Reduces unnecessary computations and Optimizes execution plans.
it improves performance for large datasets.

Q4: What is the difference between CSV and Parquet in terms of storage (row-based vs.
columnar) and why does it matter for performance?

Ans:CSV (Row-based Storage):

CSV stores data row by row.
It is simple, easy to read, and supported by many applications.
It does not store schema information and usually requires more storage space.
Spark has to read the entire file even when only a few columns are needed, making it slower for large datasets.

Parquet (Columnar Storage):

Parquet stores data column by column.
It stores schema information and uses compression, which reduces file size.
Spark reads only the required columns, making data processing faster and more efficient.
Parquet is preferred for big data and analytics because it provides better performance than CSV.

Q7: How does Spark use the Lineage Graph (DAG) to provide fault tolerance if a worker
node fails?

ans:Spark stores all transformations in a Lineage Graph (DAG). If a worker node fails, Spark uses the DAG to recompute only the lost data instead of rerunning the entire application.

Q9: Explain the concept of Predicate Pushdown in Parquet and how it affects the amount
of data loaded into memory.

ans: Predicate Pushdown means Spark sends filter conditions directly to the data source (such as Parquet).

Example:df.filter(df["Sales"] > 1000)
Only the needed data is read from disk.

It loads less data into memory and reduces I/O operations which
improves performance.

Q12: Write the Spark command to load a Parquet file from "path/to/input", filter out
any rows where user_id is null, and save the result as a CSV at "path/to/output".


```python
#remove null values
clean_df = df.filter(col("Customer ID").isNotNull())
clean_df.show(5)
```

```python
#save as csv
clean_df.write \
    .mode("overwrite") \
    .option("header", True) \
    .csv("output_csv")
#save as parquet
clean_df.write \
    .mode("overwrite") \
    .parquet("output_parquet")
```

```python
# to write and read Parquet files.
parquet_df = spark.read.parquet("output_parquet")

parquet_df.show(5)
```

Q15: When exploring a dataset, why is it safer to use .show(5) instead of .collect() on
a multi-terabyte dataset?

ans:df.show(5)
Displays only the first 5 rows
and Uses very little memory whereas
df.collect()
Brings the entire dataset to the driver machine.
It Can cause memory errors or crash the application.

Therefore, .show(5) is much safer for exploring large datasets.
